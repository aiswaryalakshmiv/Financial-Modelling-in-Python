```python
#Correlation = 0.12 + 0.24*e^(-50*PD)
#WCDR=NORM.S.DIST((NORM.S.INV(PD)+(SQRT(Correlation)*NORM.S.INV(confidence_level)))/SQRT(1-Correlation))
#Expected Loss (EL) = EAD x PD x LGD
#Stressed Loss (SL) = EAD x WCDR x LGD
#Unexpected Loss (UL) = Stressed Loss (SL) - Expected Loss (EL)
```

```python
data['Correlation']=0.12+(0.24*(np.exp(data['PD']*(-50))))
confidence_level=0.999
data['WCDR']=norm.cdf((norm.ppf(data['PD'])+(np.sqrt(data['Correlation'])*norm.ppf(confidence_level)))/(np.sqrt(1-data['Correlation'])))
data['EL']=data['EAD']*data['PD']*data['LGD']
data['SL']=data['EAD']*data['WCDR']*data['LGD']
data['UL']=data['SL']-data['EL']
data
```

```python
Expected_Loss=np.sum(data['EL'])
Stressed_Loss=np.sum(data['SL'])
Unexpected_Loss=np.sum(data['UL'])

print('Expected_Loss = $',Expected_Loss,'mn')
print('Stressed_Loss = $',Stressed_Loss,'mn')
print('Unexpected_Loss = $',Unexpected_Loss,'mn')

```
