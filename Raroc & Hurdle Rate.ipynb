```python
import pandas as pd
```

```python
import numpy as np
```

```python
from scipy.stats import norm
```

```python
data=pd.read_excel("C:/Users/balak/Anaconda Scripts/raroc.xlsx")
data
```

```python
#workings for computation
```

```python
data['EL']=data['EAD']*data['PD']*data['LGD']
data['EL']
```

```python
data['PD_variance']=data['PD']*(1-data['PD'])
data['PD_variance']
```

```python
µLGD=data['LGD'].mean()
µLGD
```

```python
data['LGD_variance']=(data['LGD']-µLGD)**2
data['LGD_variance']
```

```python
data['UL']=data['EAD']*(((data['PD']*data['LGD_variance'])+(data['PD_variance']*(data['LGD']**2)))**0.5)
data['UL']
```

```python
data['int_inc']=data['EAD']*data['Interest Rate']
data['int_inc']
```

```python
print (data)
```

```python
#Economic Capital Computation - VAR (set at bank level)
VAR=0.9997
Capital_Multiplier=norm.ppf(VAR)
Capital_Multiplier
```

```python
Economic_Capital=Capital_Multiplier*np.sum(data['UL'])
Economic_Capital
```

```python
#RAROC Computation
```

```python
Portfolio_income=np.sum(data['int_inc'])
Portfolio_income
```

```python
Interest_cost=0.09
Cost_of_fund=Interest_cost*np.sum(data['EAD'])
Cost_of_fund
```

```python
tax=0.30
```

```python
Risk_Adjusted_Return=(Portfolio_income-Cost_of_fund-(np.sum(data['EL'])))*(1-tax)
Risk_Adjusted_Return
```

```python
RAROC=Risk_Adjusted_Return/Economic_Capital
RAROC
```

```python
#Hurdle Rate Computation
```

```python
Eq_cap=3000
Pref_cap=2000
Re=0.10
Rp=0.09

Cost_of_capital=((Eq_cap*Re)+(Pref_cap*Rp))/(Eq_cap+Pref_cap)
Cost_of_capital
```

```python
Hurdle_rate=Cost_of_capital
```

```python
#Decision Maker
```

```python
if RAROC > Hurdle_rate:
    print ("Yes, Project approved")
else:
    print ("No, Project not approved")
```

```python
excess_return=RAROC-Hurdle_rate
excess_return
```

```python
if excess_return>0:
    print("Yes, Project approved")
else:
    print("No, Project not approved")
```
